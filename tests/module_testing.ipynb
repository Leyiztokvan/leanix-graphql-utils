### Test Other methods

```python

# app IDs to df
def get_app_ids() -> pd.DataFrame:
    return gqlix.get_apps_new()

def get_apps_with_no_abo() -> pd.DataFrame:
    return gqlix.get_apps_new(query_filters={"subscriptions":"[__missing__]"})

def get_apps_with_no_licence() -> pd.DataFrame:
    return gqlix.get_apps_new(query_filters={"fact_sheet_type": ["Application"], "licence":"[__missing__]"})

def get_apps_with_no_abo_and_licence() -> pd.DataFrame:
    apps_with_no_abo = get_apps_with_no_abo()
    apps_with_no_licence = get_apps_with_no_licence()
    return apps_with_no_abo.merge(apps_with_no_licence, left_on="id")

def get_apps_completion() -> pd.DataFrame:
    return gqlix.get_apps_new(on_app_nodes=["completion {percentage}"]).sort_values(by= "completion.percentage", axis=0)

def get_apps_type() -> pd.DataFrame:
    return gqlix.get_apps_new(on_app_nodes=["apptype"])
```

```python

app_ids = gqlix.get_apps_ids()
apps_completions = gqlix.get_apps_completion()
apps_broken_quality_seal = gqlix.get_apps_with_broken_quality_seal() 
apps_no_abo = gqlix.get_apps_with_no_subscription()


gqlix.save_df_as_csv(app_ids, "app_ids")
gqlix.save_df_as_csv(apps_completions, "apps_completion")
gqlix.save_df_as_csv(apps_broken_quality_seal, "apps_broken_quality_seal")
gqlix.save_df_as_csv(apps_no_abo, "apps_no_abo")
```
