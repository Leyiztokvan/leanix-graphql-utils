### FactSheet `UserGroup`

```python

filters = {"fact_sheet_type": {"keys":["UserGroup"]}, "quality_seal": {"keys":["BROKEN_QUALITY_SEAL"]}}
_user_group_nodes = ["type", "qualitySeal"]
_node_nodes = ["createdAt", "status"]

user_groups_df = gqlix.get_user_groups(facet_filters= filters, 
                                       on_user_group_nodes= _user_group_nodes, 
                                       on_node_nodes= _node_nodes).sort_values(by="level")
user_groups_df
```

```python
# get all user_groups at level 3 that have "Bank" as parent 
name = "Bank"
_bank_bank_id = user_groups_df.loc[user_groups_df["displayName"] == name, "id"]
# raise an error if name is not found
_bank_bank_id = _bank_bank_id .iloc[0]
print("Bank Id: ", _bank_bank_id)

#_bank_bank_id = "XXXX-XXXX-XXXX-XXXX-XXXX" # In case get the id (see above) did not work  

user_groups_with_parent_BANK_bank_df= gqlix.get_user_groups_with_parentX(parent_leanix_id=_bank_bank_id, level="3")
user_groups_with_parent_BANK_bank_df
```

```python
# get all user_groups at" level = 2" that have "Bank" as parent
user_groups_by_hierarchy_level_df = gqlix.get_user_groups_by_hierarchy_level(level= "2" , relation_to_parent=True, parent_leanix_id = "1ccd98cf-090f-4f04-83e6-b480702e602a")
user_groups_by_hierarchy_level_df
```
